```python
from sklearn.feature_extraction.text import CountVectorizer
import jieba

def tokenize_zh(text):
    words = jieba.lcut(text)
    return words

vectorizer = CountVectorizer(tokenizer=tokenize_zh)
```

```python
from sentence_transformers import SentenceTransformer
from keybert import KeyBERT

model = SentenceTransformer('models/paraphrase-multilingual-MiniLM-L12-v2')
kb = KeyBERT(model=model)
doc = "我爱北京天安门"
keywords = kb.extract_keywords(doc, vectorizer=vectorizer)
print(keywords)
```

```python
# 梦想家内容话术参考0905.18更新
s = '''
力证产品质量/安全性		渗透豪华外观内饰标签			传递搭载智驾绝对优势
驾控是豪华的灵魂		梦想家独创中式美学				岚图x华为，相见在顶峰
满级续航不打折			岚图，最懂消费者审美的车企		华为智驾梦想家，遥遥领先
五星最安全MPV			最符合中国审美的MPV				自研+华为，两腿大步迈向前
同级唯一魔毯空悬		梦想家，拒绝“埃尔法”			MPV科技新标杆
'''
```

```python
import re

total_kws = []
ss = re.split(r'[\s]+|\n', s)
for i in ss:
    if re.match(r'\S+', i):
        kws = kb.extract_keywords(i, vectorizer=vectorizer)
        total_kws.extend([ kw[0] for kw in kws ])
        print(i, kws)
    
print(total_kws)
```
